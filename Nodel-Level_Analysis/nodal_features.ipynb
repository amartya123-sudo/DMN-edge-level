```python
import pandas as pd
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt
import graphviz
import community
from operator import itemgetter
import os
```

```python
combined_phenotype='metadata.csv'
labels_mappings= pd.read_csv(combined_phenotype)
labels_mappings
```

```python
class local_measures_binary:
  def __init__(self, G, df):
    self.G = G
    self.df = df
    self.measures = dict()
    
    self.get_measures()
    self.measures = self.get_properties_table()

  def get_measures(self):
    deg = dict(self.G.degree(weight="weight"))
    bc = nx.betweenness_centrality(self.G, weight="weight")
    lclu = nx.clustering(self.G, weight="weight")
    andeg = nx.average_neighbor_degree(self.G, weight="weight")

    m = {
      'degree' : deg,
      'betweenness centrality': bc,
      "local clustering":lclu,
      "avg neighbour degree": andeg,
    }

    self.measures.update(m)

  def get_properties_table(self):
    return pd.DataFrame(self.measures)
```

```python
uw = './fisher-transformed/CN/'
dest_dir = './weighted_features/CN/'
```

```python
dirs = [uw]

graphs = []
labels = []

label_lookup = dict(
    zip(
        labels_mappings["subject_id"].astype(str),
        labels_mappings["group"]
    )
)

for dir in dirs:
    for file in os.listdir(dir):
        if not file.endswith("_adj.csv"):
            continue

        subject = file.replace("_adj.csv", "")

        if subject not in label_lookup:
            print(f"Subject not found: {subject}")
            continue


        if subject not in label_lookup:
            print(f"Subject not found: {subject}")
            continue

        label = label_lookup[subject]

        labels.append((label, subject))

        file_path = os.path.join(dir, file)
        df = pd.read_csv(file_path, header=None).values

        G = nx.from_numpy_array(df)
        res = local_measures_binary(G, df)  
        res_path = os.path.join(dest_dir, file)
        res.measures.to_csv(res_path, index=False)
        print("Subject",subject)
```
